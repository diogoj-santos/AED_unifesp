```python
import numpy as np

# Calculando Q1, Q3 e IQR
Q1 = df['LIMIT_BAL'].quantile(0.25)
Q3 = df['LIMIT_BAL'].quantile(0.75)
IQR = Q3 - Q1

# Definindo o limite superior para outliers
limite_superior = Q3 + 1.5 * IQR

# Encontrando todos os outliers acima do limite superior
outliers = df[df['LIMIT_BAL'] > limite_superior]

# Identificando o maior outlier
if not outliers.empty:
    maior_outlier = outliers['LIMIT_BAL'].max()

    # Removendo apenas o maior outlier
    df = df[df['LIMIT_BAL'] != maior_outlier]

# Exibir o novo boxplot sem o maior outlier
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['LIMIT_BAL'])
plt.title('Boxplot para LIMIT_BAL (Sem o Maior Outlier)')
plt.show()
```
